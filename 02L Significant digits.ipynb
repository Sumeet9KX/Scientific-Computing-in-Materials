## Patriot Missile

```python
from binary_fractions import Binary
```

```python
b_true = Binary(0.1)
print(b_true)
```

```python
b_stored = Binary("0.00011001100110011001100")
print(b_stored)
```

```python
b_error = b_true - b_stored
print(b_error)
```

```python
decimal_error_per_tick = float(b_error)
print(decimal_error_per_tick) # this error is per tick
```

```python
decimal_error_100hrs = decimal_error_per_tick*10*60*60*100
print(decimal_error_100hrs)
```

```python
velocity = 1676 # m/s
distance_error = velocity*decimal_error_100hrs
print(distance_error, "metres")
```
