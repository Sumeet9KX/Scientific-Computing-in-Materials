## There are two ways to loop over python lists:
1. to use the `for` loop and go over each element
2. Use list comprehension, i.e. a concise way of writing the `for` loop

```python
temp_c = [20, 25, 32]
# temp_k = []
# for temp in temp_c:
#     temp_k.append(temp + 273.15)
temp_k = [temp + 273.15 for temp in temp_c]
print(temp_k)
```

## Shapes of numpy arrays are important for matrix operations using @.
The following are the matrix multiplications:
1. Gives correct answer, but the shape is incorrect.
2. Results in  `error` due to shape mismatch.
3. Gives correct answer with the correct shape.

```python
# 1. Gives correct answer, but the shape is incorrect.
matrix = np.array([[1, 2], [3, 4]])
print("matrix: \n", matrix)
vector = np.array([1, 2])
print("vector: \n", vector)
print("vector.shape: ", vector.shape)
mult = matrix @ vector
print("multiplication vector: \n", mult)
print("multiplication matrix shape: ", mult.shape)
```

```python
# 2. Results in error due to shape mismatch.
matrix = np.array([[1, 2], [3, 4]])
print("matrix: \n", matrix)
vector = np.array([[1, 2]])
print("vector: \n", vector)
print("vector.shape: ", vector.shape)
mult = matrix @ vector
print("multiplication vector: \n", mult)
print("multiplication matrix shape: ", mult.shape)
```

```python
# 3. Gives correct answer with the correct shape.
matrix = np.array([[1, 2], [3, 4]])
print("matrix: \n", matrix)
vector = np.array([[1], [2]])
print("vector: \n", vector)
print("vector.shape: ", vector.shape)
mult = matrix @ vector
print("multiplication vector: \n", mult)
print("multiplication matrix shape: ", mult.shape)
```

Numpy is faster than regular python lists operations due to `Vectorization`. Always use `Vectorization` when working with arrays of numbers and avoid `for` loops.

```python
%timeit [x**2 for x in range(1000)]
```

```python
%timeit np.arange(1000)**2
```
